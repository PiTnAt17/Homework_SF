```python
connection = psycopg2.connect(
   dbname=DBNAME,
   user=USER,
   host=HOST,
   password=PASSWORD,
   port=PORT
)

query = """

WITH filtered_data AS (
    SELECT 
        a.name AS region_name,
        v.id AS vac_id
    FROM employers e
    INNER JOIN vacancies v ON v.employer_id = e.id
    INNER JOIN areas a ON a.id = v.area_id
    WHERE e.name = 'Яндекс'
        AND a.name IN ('Москва', 'Санкт-Петербург', 'Новосибирск', 'Екатеринбург', 
                       'Казань', 'Нижний Новгород', 'Челябинск', 'Самара', 
                       'Омск', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 
                       'Пермь', 'Воронеж', 'Волгоград', 'Краснодар')
),

COMBINED AS (

    SELECT 
        region_name,
        COUNT(vac_id) AS vac_count
    FROM filtered_data
    GROUP BY region_name

    UNION ALL

    SELECT 
        'Total' AS region_name,
        COUNT(vac_id) AS vac_count
    FROM filtered_data
    
            )
            
SELECT
    region_name,
    vac_count,
    CASE WHEN region_name = 'Total' THEN 1 ELSE 0 END
FROM COMBINED
ORDER BY
    vac_count DESC;
"""

df = pd.read_sql_query(query, connection)

df
```

```python
connection = psycopg2.connect(
   dbname=DBNAME,
   user=USER,
   host=HOST,
   password=PASSWORD,
   port=PORT
)

query = """
WITH ds_vacancies AS (
    SELECT 
        CASE 
            WHEN salary_from IS NOT NULL AND salary_to IS NOT NULL 
                THEN (salary_from + salary_to) / 2.0
            WHEN salary_from IS NOT NULL 
                THEN salary_from
            WHEN salary_to IS NOT NULL 
                THEN salary_to
        END AS calculated_salary,
        experience
    FROM vacancies
    WHERE (
        LOWER(name) LIKE '%data scientist%'
        OR LOWER(name) LIKE '%data science%'
        OR name ILIKE '%исследователь данных%'
        OR (LOWER(name) LIKE '%ml%' AND LOWER(name) NOT LIKE '%html%')
        OR LOWER(name) LIKE '%machine learning%'
        OR name ILIKE '%машинн%обучен%'
    )
    AND (salary_from IS NOT NULL OR salary_to IS NOT NULL)
)
SELECT 
    ROUND(AVG(calculated_salary)) AS avg_salary_3_6_years
FROM ds_vacancies
WHERE experience = 'От 3 до 6 лет'
  AND calculated_salary IS NOT NULL;
"""
df = pd.read_sql_query(query, connection)
print(df)
```
